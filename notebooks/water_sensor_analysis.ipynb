```python
# Look at consecutive overflow readings for each sensor

overflow_df = water_df[
    water_df["overflow_flag"] == 1
].copy()

overflow_df = overflow_df.sort_values(
    ["sensor_id", "timestamp"]
)

overflow_df["time_gap"] = (
    overflow_df.groupby("sensor_id")["timestamp"].diff()
)

print("Time gaps between consecutive overflow records:")
print(overflow_df["time_gap"].value_counts().head(10))
```

```python
# Sort the complete dataset first
water_df = water_df.sort_values(
    ["sensor_id", "timestamp"]
).reset_index(drop=True)

# Create a new event group whenever overflow changes
water_df["overflow_change"] = (
    water_df.groupby("sensor_id")["overflow_flag"]
    .transform(lambda x: x.ne(x.shift()).cumsum())
)

# Keep only overflow records
continuous_overflow = water_df[
    water_df["overflow_flag"] == 1
].copy()

# Calculate duration of each continuous overflow event
event_duration = (
    continuous_overflow
    .groupby(["sensor_id", "overflow_change"])
    .agg(
        start_time=("timestamp", "min"),
        end_time=("timestamp", "max"),
        readings=("timestamp", "count")
    )
    .reset_index()
)

# Duration in minutes
event_duration["duration_minutes"] = (
    event_duration["readings"] * 15
)

print("Number of continuous overflow events:")
print(len(event_duration))

print("\nOverflow event duration:")
print(
    event_duration["duration_minutes"]
    .value_counts()
    .sort_index()
    .head(20)
)
```

```python
event_duration["duration_minutes"] = (
    event_duration["end_time"] - event_duration["start_time"]
).dt.total_seconds() / 60

print("Corrected overflow event durations:")
print(
    event_duration["duration_minutes"]
    .value_counts()
    .sort_index()
)
```

```python
# Get all overflow records
overflow_rows = water_df[
    water_df["overflow_flag"] == 1
].copy()

# Look at the sensor conditions during overflow
overflow_summary = overflow_rows[
    [
        "water_level_cm",
        "rise_rate_cm_per_15min",
        "rainfall_mm_per_hr",
        "cumulative_rain_6hr_mm",
        "cumulative_rain_24hr_mm"
    ]
].describe()

print("Conditions during overflow:")
display(overflow_summary)
```

```python
# Make sure data is sorted correctly
water_df = water_df.sort_values(
    ["sensor_id", "timestamp"]
).reset_index(drop=True)

# Future overflow target: overflow in the next 15 minutes
water_df["overflow_next_15min"] = (
    water_df.groupby("sensor_id")["overflow_flag"]
    .shift(-1)
)

print("Future target distribution:")
print(
    water_df["overflow_next_15min"]
    .value_counts(dropna=False)
)
```

```python
# Remove rows where future target is unavailable
target_df = water_df.dropna(
    subset=["overflow_next_15min"]
).copy()

print("Usable rows:", len(target_df))

print("\nFuture overflow distribution:")
print(
    target_df["overflow_next_15min"]
    .value_counts()
)

print("\nFuture overflow percentage:")
print(
    target_df["overflow_next_15min"]
    .value_counts(normalize=True).mul(100).round(2)
)
```

```python
# Compare current overflow with overflow in the next 15 minutes

transition_table = pd.crosstab(
    water_df["overflow_flag"],
    water_df["overflow_next_15min"],
    dropna=False
)

print("Current overflow → next 15-minute overflow:")
display(transition_table)
```

```python
water_df["overflow_onset_next_15min"] = np.where(
    (water_df["overflow_flag"] == 0) &
    (water_df["overflow_next_15min"] == 1),
    1,
    0
)
```

```python
water_df["overflow_onset_next_15min"] = np.nan

valid_transition = (
    water_df["overflow_flag"].notna() &
    water_df["overflow_next_15min"].notna()
)

water_df.loc[valid_transition, "overflow_onset_next_15min"] = (
    (
        (water_df.loc[valid_transition, "overflow_flag"] == 0) &
        (water_df.loc[valid_transition, "overflow_next_15min"] == 1)
    ).astype(int)
)

print(
    water_df["overflow_onset_next_15min"]
    .value_counts(dropna=False)
)
```

```python
feature_columns = [
    "water_level_cm",
    "rise_rate_cm_per_15min",
    "rainfall_mm_per_hr",
    "cumulative_rain_6hr_mm",
    "cumulative_rain_24hr_mm",
    "pipe_diameter_mm",
    "pipe_material",
    "pipe_slope",
    "elevation_m",
    "month",
    "overflow_onset_next_15min"
]

model_df = water_df[feature_columns].copy()

print("Model dataset shape:")
print(model_df.shape)

print("\nColumns:")
print(model_df.columns.tolist())
```

```python
model_df = model_df.dropna(
    subset=["overflow_onset_next_15min"]
).copy()

print("Model dataset after removing missing targets:")
print(model_df.shape)

print("\nMissing values:")
print(model_df.isnull().sum())
```

```python
# Add timestamp back to the clean modeling dataset
model_df["timestamp"] = water_df.loc[
    model_df.index, "timestamp"
]

# Sort chronologically
model_df = model_df.sort_values(
    "timestamp"
).reset_index(drop=True)

print("Model dataset shape:")
print(model_df.shape)

print("\nTime range:")
print("Start:", model_df["timestamp"].min())
print("End:", model_df["timestamp"].max())
```

```python
# 80% chronological split

split_index = int(len(model_df) * 0.80)

train_df = model_df.iloc[:split_index].copy()
test_df = model_df.iloc[split_index:].copy()

print("Training data:")
print(train_df.shape)

print("\nTesting data:")
print(test_df.shape)

print("\nTraining time:")
print(train_df["timestamp"].min(), "to", train_df["timestamp"].max())

print("\nTesting time:")
print(test_df["timestamp"].min(), "to", test_df["timestamp"].max())
```

```python
print("Training target distribution:")
print(
    train_df["overflow_onset_next_15min"]
    .value_counts()
)

print("\nTraining target percentage:")
print(
    train_df["overflow_onset_next_15min"]
    .value_counts(normalize=True).mul(100).round(3)
)

print("\nTesting target distribution:")
print(
    test_df["overflow_onset_next_15min"]
    .value_counts()
)

print("\nTesting target percentage:")
print(
    test_df["overflow_onset_next_15min"]
    .value_counts(normalize=True).mul(100).round(3)
)
```

```python
monthly_onset = (
    model_df[model_df["overflow_onset_next_15min"] == 1]
    .groupby("timestamp")
    .size()
)

# Convert timestamps to month
monthly_counts = (
    model_df[model_df["overflow_onset_next_15min"] == 1]
    .assign(month=lambda x: x["timestamp"].dt.month)
    .groupby("month")
    .size()
)

print("New overflow events by month:")
print(monthly_counts)
```

```python
# Check proposed chronological periods

train_period = model_df[
    model_df["timestamp"] < "2026-08-01"
]

validation_period = model_df[
    (model_df["timestamp"] >= "2026-08-01") &
    (model_df["timestamp"] < "2026-09-01")
]

test_period = model_df[
    model_df["timestamp"] >= "2026-09-01"
]

print("TRAIN: Jan-Jul")
print("Rows:", len(train_period))
print("Positive events:",
      (train_period["overflow_onset_next_15min"] == 1).sum())

print("\nVALIDATION: August")
print("Rows:", len(validation_period))
print("Positive events:",
      (validation_period["overflow_onset_next_15min"] == 1).sum())

print("\nTEST: Sep-Dec")
print("Rows:", len(test_period))
print("Positive events:",
      (test_period["overflow_onset_next_15min"] == 1).sum())
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_period["overflow_onset_next_15min"]
)

class_weight_dict = dict(
    zip(classes, class_weights)
)

print("Class weights:")
print(class_weight_dict)
```

```python
feature_columns = [
    "water_level_cm",
    "rise_rate_cm_per_15min",
    "rainfall_mm_per_hr",
    "cumulative_rain_6hr_mm",
    "cumulative_rain_24hr_mm",
    "pipe_diameter_mm",
    "pipe_material",
    "pipe_slope",
    "elevation_m",
    "month"
]

target_column = "overflow_onset_next_15min"

X_train = train_period[feature_columns].copy()
y_train = train_period[target_column].copy()

X_validation = validation_period[feature_columns].copy()
y_validation = validation_period[target_column].copy()

X_test = test_period[feature_columns].copy()
y_test = test_period[target_column].copy()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_validation:", X_validation.shape)
print("y_validation:", y_validation.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Fit ONLY on training data
pipe_train = encoder.fit_transform(
    X_train[["pipe_material"]]
)

pipe_validation = encoder.transform(
    X_validation[["pipe_material"]]
)

pipe_test = encoder.transform(
    X_test[["pipe_material"]]
)

print("Encoded categories:")
print(encoder.categories_)
```

```python
# Numerical features
numeric_features = [
    "water_level_cm",
    "rise_rate_cm_per_15min",
    "rainfall_mm_per_hr",
    "cumulative_rain_6hr_mm",
    "cumulative_rain_24hr_mm",
    "pipe_diameter_mm",
    "pipe_slope",
    "elevation_m",
    "month"
]

# Convert numerical features to NumPy arrays
X_train_numeric = X_train[numeric_features].to_numpy()
X_validation_numeric = X_validation[numeric_features].to_numpy()
X_test_numeric = X_test[numeric_features].to_numpy()

# Combine numerical + one-hot encoded pipe material
X_train_final = np.hstack([
    X_train_numeric,
    pipe_train
])

X_validation_final = np.hstack([
    X_validation_numeric,
    pipe_validation
])

X_test_final = np.hstack([
    X_test_numeric,
    pipe_test
])

print("Final training features:", X_train_final.shape)
print("Final validation features:", X_validation_final.shape)
print("Final test features:", X_test_final.shape)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

# Different percentages of the chronological training data
train_sizes = [0.20, 0.40, 0.60, 0.80, 1.00]

train_f1_scores = []
validation_f1_scores = []

for fraction in train_sizes:

    # Take the first part of the chronological training data
    n_samples = int(len(X_train_final) * fraction)

    X_train_subset = X_train_final[:n_samples]
    y_train_subset = y_train.iloc[:n_samples]

    # Create a new Random Forest
    model = RandomForestClassifier(
        n_estimators=200,
        max_depth=12,
        min_samples_leaf=5,
        class_weight=class_weight_dict,
        random_state=42,
        n_jobs=-1
    )

    print(f"Training with {fraction*100:.0f}% data ({n_samples} samples)...")

    model.fit(X_train_subset, y_train_subset)

    # -------------------------
    # Training prediction
    # -------------------------
    y_train_pred = model.predict(X_train_subset)

    train_f1 = f1_score(
        y_train_subset,
        y_train_pred,
        zero_division=0
    )

    # -------------------------
    # Validation prediction
    # -------------------------
    y_val_pred = model.predict(X_validation_final)

    val_f1 = f1_score(
        y_validation,
        y_val_pred,
        zero_division=0
    )

    train_f1_scores.append(train_f1)
    validation_f1_scores.append(val_f1)

    print(
        f"Train F1: {train_f1:.4f} | "
        f"Validation F1: {val_f1:.4f}"
    )

# -------------------------
# Plot learning curve
# -------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    np.array(train_sizes) * 100,
    train_f1_scores,
    marker="o",
    label="Training F1"
)

plt.plot(
    np.array(train_sizes) * 100,
    validation_f1_scores,
    marker="o",
    label="Validation F1"
)

plt.xlabel("Training Data Used (%)")
plt.ylabel("F1 Score")
plt.title("Random Forest Learning Curve")
plt.xticks([20, 40, 60, 80, 100])
plt.grid(True)
plt.legend()

plt.show()
```

```python
from sklearn.metrics import f1_score

# precision and recall have one more element than thresholds
f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_idx = np.argmax(f1_scores)

best_threshold = thresholds[best_idx]
best_precision = precision[best_idx]
best_recall = recall[best_idx]
best_f1 = f1_scores[best_idx]

print("Best threshold based on F1")
print("--------------------------")
print(f"Threshold : {best_threshold:.4f}")
print(f"Precision : {best_precision:.4f}")
print(f"Recall    : {best_recall:.4f}")
print(f"F1 Score  : {best_f1:.4f}")
```

```python
from sklearn.metrics import confusion_matrix

selected_thresholds = [0.50, 0.70, 0.80, 0.90, 0.95, 0.9656, 0.98]

threshold_analysis = []

for threshold in selected_thresholds:

    y_pred = (y_val_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_validation,
        y_pred
    ).ravel()

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = (
        2 * precision * recall / (precision + recall)
        if (precision + recall) > 0 else 0
    )

    threshold_analysis.append({
        "Threshold": threshold,
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "TN": tn,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    })

threshold_analysis_df = pd.DataFrame(threshold_analysis)

threshold_analysis_df
```

```python
# Freeze threshold selected from validation
final_threshold = 0.9656

# Test probabilities
y_test_prob = rf_model.predict_proba(X_test_final)[:, 1]

# Test predictions
y_test_pred = (y_test_prob >= final_threshold).astype(int)

# Evaluate
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score
)

test_precision = precision_score(
    y_test,
    y_test_pred,
    zero_division=0
)

test_recall = recall_score(
    y_test,
    y_test_pred,
    zero_division=0
)

test_f1 = f1_score(
    y_test,
    y_test_pred,
    zero_division=0
)

test_pr_auc = average_precision_score(
    y_test,
    y_test_prob
)

test_roc_auc = roc_auc_score(
    y_test,
    y_test_prob
)

print("FINAL TEST RESULTS")
print("==================")
print(f"Threshold : {final_threshold}")
print(f"Precision : {test_precision:.4f}")
print(f"Recall    : {test_recall:.4f}")
print(f"F1 Score  : {test_f1:.4f}")
print(f"PR-AUC    : {test_pr_auc:.4f}")
print(f"ROC-AUC   : {test_roc_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_test_pred,
        digits=4,
        zero_division=0
    )
)
```

```python
import numpy as np

def predict_overflow(
    water_level_cm,
    rise_rate_cm_per_15min,
    rainfall_mm_per_hr,
    cumulative_rain_6hr_mm,
    cumulative_rain_24hr_mm,
    pipe_diameter_mm,
    pipe_material,
    pipe_slope,
    elevation_m,
    month
):

    # 9 numerical features
    numeric_data = np.array([[
        water_level_cm,
        rise_rate_cm_per_15min,
        rainfall_mm_per_hr,
        cumulative_rain_6hr_mm,
        cumulative_rain_24hr_mm,
        pipe_diameter_mm,
        pipe_slope,
        elevation_m,
        month
    ]])

    # Encode pipe material
    categorical_data = encoder.transform(
        [[pipe_material]]
    )

    # Combine -> 12 features
    final_input = np.hstack([
        numeric_data,
        categorical_data
    ])

    print("Input shape:", final_input.shape)
    print("Model expects:", model.n_features_in_)

    # Probability
    probability = model.predict_proba(
        final_input
    )[0, 1]

    # Threshold
    prediction = probability >= threshold

    print("\n-------------------------------")
    print(" FloodSentinels Prediction")
    print("-------------------------------")

    print(f"Probability : {probability:.4f}")
    print(f"Threshold   : {threshold:.4f}")

    if prediction:
        print("Prediction  : YES ⚠️")
        print("Overflow onset likely in next 15 minutes.")
    else:
        print("Prediction  : NO ✅")
        print("No overflow onset predicted in next 15 minutes.")
```
