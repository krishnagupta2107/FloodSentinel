```python
year_counts = (
    historical_df
    .groupby("SUBDIVISION")["YEAR"]
    .nunique()
    .sort_values()
)

year_counts
```

```python
year_counts.value_counts().sort_index()
```

```python
incomplete_subdivisions = year_counts[year_counts < 115]

print("Subdivisions with incomplete year records:")
print(incomplete_subdivisions)
```

```python
year_counts < 115
```

```python
all_years = set(range(1901, 2016))

for subdivision in incomplete_subdivisions.index:
    available_years = set(
        historical_df.loc[
            historical_df["SUBDIVISION"] == subdivision,
            "YEAR"
        ]
    )

    missing_years = sorted(all_years - available_years)

    print("\nSubdivision:", subdivision)
    print("Missing years:", missing_years)
```

```python
historical_missing_percent = (
    historical_df.isnull().mean() * 100
).round(2)

historical_missing_percent
```

```python
rainfall_columns = [
    "JAN", "FEB", "MAR", "APR", "MAY", "JUN",
    "JUL", "AUG", "SEP", "OCT", "NOV", "DEC",
    "ANNUAL", "Jan-Feb", "Mar-May", "Jun-Sep", "Oct-Dec"
]

negative_counts = (historical_df[rainfall_columns] < 0).sum()

negative_counts
```

```python
district_negative_counts = (district_df[rainfall_columns] < 0).sum()

district_negative_counts
```

```python
historical_df[rainfall_columns].describe()
```

```python
district_df[rainfall_columns].describe()
```

```python
Q1 = historical_df[rainfall_columns].quantile(0.25)
Q3 = historical_df[rainfall_columns].quantile(0.75)

IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

outlier_counts = (
    historical_df[rainfall_columns] > upper_bound
).sum()

outlier_counts
```

```python
monthly_sum = historical_df[
    ["JAN", "FEB", "MAR", "APR", "MAY", "JUN",
     "JUL", "AUG", "SEP", "OCT", "NOV", "DEC"]
].sum(axis=1, min_count=1)

difference = historical_df["ANNUAL"] - monthly_sum

difference.describe()
```
